```python
import sys
import os
import json

# Add the parent directory to the Python path to import the 'appraisal' module
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from appraisal.evaluator import run_all_trees
print("Successfully imported the appraisal module!")
```

```python
# Load the sample paper text
paper_path = os.path.join(project_root, "data/sample_papers/sample_papers_md/1.md")

with open(paper_path, "r", encoding="utf-8") as f:
    paper_text = f.read()
    
print(f"Loaded paper '{os.path.basename(paper_path)}'. Length: {len(paper_text)} characters.")
```

```python
# Load the CEECAT guidance text
guidance_path = os.path.join(project_root, "data/guidance_intro_md/guidance_intro.md")
guidance_text = None

if os.path.exists(guidance_path):
    with open(guidance_path, "r", encoding="utf-8") as f:
        guidance_text = f.read()
    print("Loaded CEECAT guidance text.")
else:
    print(f"Warning: Guidance text not found at {guidance_path}")
```

```python
# Run the full appraisal decision tree process
print("Starting appraisal on all decision trees...\n")
results = run_all_trees(paper_text=paper_text, guidance_text=guidance_text, verbose=True)
```

```python
# Display the final results
print("\n--- Final Appraisal Results ---")
print(json.dumps(results, indent=2))
```
