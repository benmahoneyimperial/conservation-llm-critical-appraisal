```python
# Run few_shot mode on one paper and save the output JSON.
import json
import os
import sys
from pathlib import Path

repo_root = Path.cwd().resolve()
if repo_root.name == "notebooks":
    repo_root = repo_root.parent
sys.path.append(str(repo_root))

from domain_shot.build_question_scoring_prompt import (
    DEFAULT_PAPERS_DIR,
    build_prompt_messages,
    load_default_guidance_intro,
    load_domain_questions,
    read_text_file,
 )
from domain_shot.evaluation import DEFAULT_MODEL, call_llm_and_process

paper_filename = "2.md"
paper_path = os.path.join(DEFAULT_PAPERS_DIR, paper_filename)
paper_text = read_text_file(paper_path)
intro_text = load_default_guidance_intro()

results_few_shot = {}
for domain_idx in range(1, 8):
    domain_name = f"domain_{domain_idx}"
    domain_questions_text = load_domain_questions(domain_name)
    messages = build_prompt_messages(
        domain_name=domain_name,
        intro_text=intro_text,
        domain_questions_text=domain_questions_text,
        paper_text=paper_text,
    )
    results_few_shot[domain_name] = call_llm_and_process(domain_name, messages, DEFAULT_MODEL)

output_dir = repo_root / "output" / "three_prompt_single_paper" / "single_paper"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "2_few_shot.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(results_few_shot, f, indent=2, ensure_ascii=False)

print(f"Saved: {output_path}")
```

```python
# Run few_shot mode on all 10 sample papers and save one JSON file.
import json
import sys
from pathlib import Path

repo_root = Path.cwd().resolve()
if repo_root.name == "notebooks":
    repo_root = repo_root.parent
sys.path.append(str(repo_root))

from domain_shot.build_question_scoring_prompt import (
    build_prompt_messages,
    load_default_guidance_intro,
    load_domain_questions,
    read_text_file,
 )
from domain_shot.evaluation import DEFAULT_MODEL, call_llm_and_process

papers_dir = repo_root / "data" / "sample_papers" / "sample_papers_md"
paper_files = sorted(
    [p for p in papers_dir.glob("*.md")],
    key=lambda p: int(p.stem) if p.stem.isdigit() else p.stem,
 )

intro_text = load_default_guidance_intro()
results_few_shot_all = {}

for paper_file in paper_files:
    paper_text = read_text_file(str(paper_file))
    domain_results = {}
    for domain_idx in range(1, 8):
        domain_name = f"domain_{domain_idx}"
        domain_questions_text = load_domain_questions(domain_name)
        messages = build_prompt_messages(
            domain_name=domain_name,
            intro_text=intro_text,
            domain_questions_text=domain_questions_text,
            paper_text=paper_text,
        )
        domain_results[domain_name] = call_llm_and_process(domain_name, messages, DEFAULT_MODEL)
    results_few_shot_all[paper_file.name] = domain_results

output_dir = repo_root / "output" / "three_prompt_single_paper" / "all_papers"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "few_shot_all_papers.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(results_few_shot_all, f, indent=2, ensure_ascii=False)

print(f"Saved: {output_path}")
```
