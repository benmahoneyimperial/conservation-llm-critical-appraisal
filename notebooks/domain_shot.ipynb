```python
import sys
import os
import json

# Add the parent directory to the Python path to import the 'domain_shot' module
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from domain_shot.domain_shot_evaluator import load_domain_guidance, evaluate_all
print("Successfully imported evaluate_all!")
```

```python
# Load the domain guidance and introduction text
guidance_dir = os.path.join(project_root, "data/guidance_by_domain_md")
intro_file_path = os.path.join(project_root, "data/guidance_intro_md/guidance_intro.md")

guidance_dict = load_domain_guidance(guidance_dir)
with open(intro_file_path, "r", encoding="utf-8") as f:
    intro_text = f.read()

print(f"Loaded {len(guidance_dict)} domains and introduction text.")
```

```python
# Set the path to the directory containing all papers
papers_dir = os.path.join(project_root, "data/sample_papers/sample_papers_md")
print(f"Papers directory set to: {papers_dir}")
```

```python
# Run the batch domain-shot evaluation on all papers
print("Starting batch evaluation on all papers...\n")
all_results = evaluate_all(
    guidance_dict=guidance_dict,
    papers_dir=papers_dir,
    intro_text=intro_text
)
```

```python
# Display the final summary of results
print("\n--- Final Batch Appraisal Results ---")
for paper, result in all_results.items():
    if isinstance(result, dict) and 'overall_risk' in result:
        print(f"{paper}: {result['overall_risk']}")
    else:
        print(f"{paper}: Evaluation Failed or Incomplete")

# Save the detailed results to a JSON file
output_path = os.path.join(project_root, "domain_shot_results_batch.json")
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(all_results, f, indent=2)
print(f"\nDetailed results saved to {output_path}")
```
