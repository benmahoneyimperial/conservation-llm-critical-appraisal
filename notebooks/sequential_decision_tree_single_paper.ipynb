```python
# Setup and imports.
import json
import os
import sys
from pathlib import Path

repo_root = Path.cwd().resolve()
if repo_root.name == "notebooks":
    repo_root = repo_root.parent
sys.path.append(str(repo_root))

from sequential_decision_tree.evaluator import run_all_trees
```

```python
# Choose and load one paper.
PAPER_FILENAME = os.getenv("PAPER_FILENAME", "1.md")
paper_path = repo_root / "data" / "sample_papers" / "sample_papers_md" / PAPER_FILENAME

with open(paper_path, "r", encoding="utf-8") as f:
    paper_text = f.read()

print(f"Loaded {paper_path.name}")
```

```python
# Load guidance text.
guidance_path = repo_root / "data" / "guidance_intro_md" / "guidance_intro.md"
guidance_text = None
if guidance_path.exists():
    with open(guidance_path, "r", encoding="utf-8") as f:
        guidance_text = f.read()

print(f"Guidance loaded: {guidance_text is not None}")
```

```python
# Run sequential decision trees on the paper.
results = run_all_trees(
    paper_text=paper_text,
    guidance_text=guidance_text,
    verbose=True,
 )

print("Finished.")
```

```python
# Save and preview results.
output_dir = repo_root / "output" / "sequential_decision_tree_single_paper"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / f"{paper_path.stem}.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

print(f"Saved: {output_path}")
print(json.dumps(results, indent=2)[:2000])
```
