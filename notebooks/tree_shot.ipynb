```python
import sys
import os
import json

# Add the parent directory to the Python path to import the 'appraisal' module
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from appraisal.evaluator import analyse_all_papers
print("Successfully imported analyse_all_papers!")
```

```python
# Set up the paths to your papers and guidance
papers_dir = os.path.join(project_root, "data/sample_papers/sample_papers_md")
guidance_path = os.path.join(project_root, "data/guidance_intro_md/guidance_intro.md")

print(f"Papers directory: {papers_dir}")
print(f"Guidance file: {guidance_path}")
```

```python
# Run the full appraisal decision tree process on all papers in the directory
print("Starting batch appraisal...\n")
all_results = analyse_all_papers(processed_dir=papers_dir, guidance_path=guidance_path, verbose=True)
```

```python
# Display the final summary of results
print("\n--- Final Batch Appraisal Results ---")
for paper, result in all_results.items():
    if isinstance(result, dict) and 'overall' in result:
        print(f"{paper}: {result['overall']}")
    else:
        print(f"{paper}: {result}")

# Save the detailed results to a JSON file
output_path = os.path.join(project_root, "results_batch_2.json")
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(all_results, f, indent=2)
print(f"\nDetailed results saved to {output_path}")
```
