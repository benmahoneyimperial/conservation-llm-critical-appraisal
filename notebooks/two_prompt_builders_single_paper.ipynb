```python
# Build and print the tree_questions prompt for one paper.
import os
import sys
from pathlib import Path

repo_root = Path.cwd().resolve()
if repo_root.name == "notebooks":
    repo_root = repo_root.parent
sys.path.append(str(repo_root))

from domain_shot.build_tree_questions_prompt import (
    DEFAULT_PAPERS_DIR,
    build_prompt,
    load_default_guidance_intro,
    load_domain_decision_tree,
    load_domain_questions,
    read_text_file,
)

domain_name = "domain_1"
paper_filename = "2.md"
paper_path = os.path.join(DEFAULT_PAPERS_DIR, paper_filename)

intro_text = load_default_guidance_intro()
decision_tree_text = load_domain_decision_tree(domain_name)
domain_questions_text = load_domain_questions(domain_name)
paper_text = read_text_file(paper_path)

prompt_tree_questions = build_prompt(
    domain_name=domain_name,
    intro_text=intro_text,
    decision_tree_text=decision_tree_text,
    domain_questions_text=domain_questions_text,
    paper_text=paper_text,
)

print(prompt_tree_questions)
```
